```python
import random

def sample_seq(max_len=6):
    n = random.randint(1, max_len-1)
    return ["a"] * n + ["b"]

data = [sample_seq() for _ in range(2000)]
data
```

```python
eos_and_bos_seq=[['bos']+seq+['eos'] for seq in data]
eos_and_bos_seq
```

```python
l_to_ix={'a':0,
         'b':1,
         'bos':2,
         'eos':3}

import torch
import torch.nn as nn
class SmallTransformerModel(nn.Module):
    def __init__(self):
        super(SmallTransformerModel, self).__init__()
        self.d=torch.tensor(12)
        #One embedding for a and b, 2 for eos and bos
        self.embedding = nn.Embedding(4, self.d)
        self.Q=nn.Linear(self.d,self.d)
        self.K=nn.Linear(self.d,self.d)
        self.V=nn.Linear(self.d,self.d)
        self.MLP1=nn.ReLU(nn.Linear(self.d, self.d))
        self.MLP2=nn.ReLU(nn.Linear(self.d,self.d))



    def forward(self, sequence):
        sequence_emb=torch.tensor([l_to_ix[letter] for letter in sequence])    
        emb=self.embedding(torch.tensor(sequence_emb))
        q=self.Q(emb)
        k=self.K(emb)
        v=self.V(emb)
        print(q)
        attention=torch.softmax((q@k.T)/torch.sqrt(self.d),dim=1)@v
        print(attention.shape)
        mlp_1=self.MLP1(attention)
        mlp_2=self.MLP2(mlp_1)
        print(mlp_2)


model=SmallTransformerModel()
model(eos_and_bos_seq[0])
```
